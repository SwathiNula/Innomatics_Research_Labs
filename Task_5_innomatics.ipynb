```python
# Problem 1: Social Media – Post Engagement Analyzer

def analyze_post_engagement(likes_list):
    total_likes = 0

    # loop to calculate total likes
    for likes in likes_list:
        total_likes += likes

    # condition to check viral or not
    if total_likes >= 1000:
        status = "Viral Post"
    else:
        status = "Normal Engagement"

    print("Total Likes:", total_likes)
    print("Post Status:", status)

# Example input
likes = [200, 150, 300, 250, 150]

# Function call
analyze_post_engagement(likes)
```

```python
# Problem 2: Healthcare – Medicine Stock Alert System

def check_medicine_stock(stock):
    if stock < 10:
        status = "Low Stock Alert"
    else:
        status = "Stock Sufficient"

    print("Medicine Stock:", stock)
    print("Status:", status)

# Example input
medicine_stock = 6

# Function call
check_medicine_stock(medicine_stock)
```

```python
# Problem 3: Agriculture – Rainfall Adequacy Checker

def check_rainfall(rainfall_list, required_level):
    total = 0

    # loop to calculate total rainfall
    for r in rainfall_list:
        total += r

    average_rainfall = total / len(rainfall_list)

    if average_rainfall >= required_level:
        status = "Adequate Rainfall"
    else:
        status = "Inadequate Rainfall"

    print("Average Rainfall:", average_rainfall)
    print("Rainfall Status:", status)

# Example input
rainfall_data = [60, 70, 80, 78]
required = 70

# Function call
check_rainfall(rainfall_data, required)
```

```python
# Problem 4: Social Media – Duplicate Account Detection

def check_duplicate_accounts(usernames):
    unique_users = set(usernames)

    if len(unique_users) != len(usernames):
        print("Duplicate Accounts Found: Yes")
    else:
        print("Duplicate Accounts Found: No")

# Example input
users = ["john", "alice", "bob", "john"]

# Function call
check_duplicate_accounts(users)
```

```python
# Problem 6: Agriculture – Premium Crop Price Filter

def filter_premium_crops(prices):
    premium_crops = []

    for price in prices:
        if price > 2000:
            premium_crops.append(price)

    print("Premium Crops:", premium_crops)

# Example input (kept such that only two values are premium)
crop_prices = [1500, 2500, 1800, 3200, 1900]

# Function call
filter_premium_crops(crop_prices)
```

```python
# Problem 7: System Monitoring – Application Health Checker

def check_system_health(error_count):
    if error_count == 0:
        status = "Healthy"
    elif error_count <= 5:
        status = "Minor Issues"
    else:
        status = "Critical Issues"

    print("Error Count:", error_count)
    print("System Status:", status)

# Example input
errors = 7

# Function call
check_system_health(errors)
```

```python
# Problem 8: Banking – Daily Transaction Limit Checker

def check_transaction_limit(amount):
    daily_limit = 50000

    if amount <= daily_limit:
        status = "Approved"
    else:
        status = "Rejected"

    print("Transaction Amount:", amount)
    print("Transaction Status:", status)

# Example input
transaction_amount = 60000

# Function call
check_transaction_limit(transaction_amount)
```

```python
# Problem 9: E-Learning – Student Attendance Eligibility System

def check_attendance_eligibility(attendance_list):
    total_classes = len(attendance_list)
    attended_classes = 0

    # count attended classes (1 means present, 0 means absent)
    for day in attendance_list:
        if day == 1:
            attended_classes += 1

    attendance_percentage = (attended_classes / total_classes) * 100

    if attendance_percentage >= 75:
        status = "Eligible"
    else:
        status = "Not Eligible"

    print("Attendance Percentage:", attendance_percentage)
    print("Exam Eligibility:", status)

# Example input (1 = present, 0 = absent)
attendance_data = [1, 1, 1, 1, 0]

# Function call
check_attendance_eligibility(attendance_data)
```

```python
# Problem 10: Smart Electricity Bill Analyzer

def calculate_electricity_bill(units):
    bill = 0

    # slab calculation
    if units <= 100:
        bill = units * 3
    elif units <= 200:
        bill = (100 * 3) + ((units - 100) * 5)
    else:
        bill = (100 * 3) + (100 * 5) + ((units - 200) * 7)

    # usage classification
    if bill < 500:
        usage = "Low Usage"
    elif bill <= 1500:
        usage = "Moderate Usage"
    else:
        usage = "High Usage"

    return bill, usage

# Example input
units_consumed = 250

# Function call
total_bill, usage_status = calculate_electricity_bill(units_consumed)

print("Units Consumed:", units_consumed)
print("Total Bill:", total_bill)
print("Usage Status:", usage_status)
```
