```python
#Problem-1

employees = {
    "Ravi": 92,
    "Anita": 88,
    "Kiran": 92,
    "Suresh": 85
}

# find highest score
highest = 0
for name in employees:
    if employees[name] > highest:
        highest = employees[name]

# find all employees with highest score
top_people = []
for name in employees:
    if employees[name] == highest:
        top_people.append(name)

print("Top Performers Eligible for Bonus:", ", ".join(top_people), "(Score:", highest, ")")
```

```python
#Problem-2

text = "Buy mobile phone buy phone online"

# convert to lowercase
text = text.lower()

# split words
words = text.split()

count = {}

for w in words:
    if w in count:
        count[w] = count[w] + 1
    else:
        count[w] = 1

# print only repeated words
result = {}
for w in count:
    if count[w] > 1:
        result[w] = count[w]

print(result)
```

```python
#Problem-3

sensor_readings = [3, 4, 7, 8, 10, 12, 5]

valid = []

for i in range(len(sensor_readings)):
    if sensor_readings[i] % 2 == 0:
        valid.append((i, sensor_readings[i]))

print("Valid Sensor Readings (Hour, Value):")
print(valid)
```

```python
#Problem-4

emails = [
    "ravi@gmail.com",
    "anita@yahoo.com",
    "kiran@gmail.com",
    "suresh@gmail.com",
    "meena@yahoo.com"
]

domain_count = {}

for mail in emails:
    domain = mail.split("@")[1]
    if domain in domain_count:
        domain_count[domain] += 1
    else:
        domain_count[domain] = 1

total = len(emails)

for domain in domain_count:
    percent = (domain_count[domain] / total) * 100
    print(domain + ":", int(percent), "%")
```

```python
# Problem-5

sales = [1200, 1500, 900, 2200, 1400, 3000]

total = 0
for s in sales:
    total += s

average = total / len(sales)

limit = average + (average * 0.30)

for i in range(len(sales)):
    if sales[i] >= limit - 10:
        print("Day", i + 1, ":", sales[i])
```
