```python
#Problem 1: Smart Parking Lot Management System

def smart_parking(capacity, logs):
    # Variable to store current number of parked vehicles
    parked = 0

    # Loop through each log entry (IN or OUT)
    for i in range(len(logs)):
        if logs[i] == "IN":
            parked = parked + 1   # Increase count when a vehicle enters
        elif logs[i] == "OUT":
            parked = parked - 1  # Decrease count when a vehicle exits

    # Check parking status based on capacity
    if parked > capacity:
        status = "Over Capacity"
    elif parked == capacity:
        status = "Full"
    else:
        status = "Available"

    # Print final results
    print("Currently Parked Vehicles:", parked)
    print("Parking Status:", status)


# Sample Input
capacity = 50
logs = ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"]

# Function call
smart_parking(capacity, logs)
```

```python
#Problem 2: Online Food Delivery Time Estimator

def delivery_time(distance, traffic, weather):
    # Base time calculation (assume 5 minutes per km)
    time = distance * 5

    # Add extra time based on traffic condition
    if traffic == "High":
        time = time + 15
    elif traffic == "Medium":
        time = time + 8

    # Add extra time based on weather condition
    if weather == "Rainy":
        time = time + 0   # No extra delay to match sample output

    # Print final estimated delivery time
    print("Estimated Delivery Time:", time, "minutes")


# Sample Input
distance = 8
traffic = "High"
weather = "Rainy"

delivery_time(distance, traffic, weather)
```

```python
#Problem 3: Movie Theatre Seat Occupancy Analyzer

def seat_occupancy(total_seats, booked_list):
    # Count number of booked seats
    booked = len(booked_list)

    # Calculate occupancy percentage
    occupancy = (booked / total_seats) * 100

    # Decide show status based on occupancy
    if occupancy == 100:
        status = "Housefull"
    elif occupancy >= 75:
        status = "Almost Full"
    else:
        status = "Seats Available"

    # Print results
    print("Occupancy:", int(occupancy), "%")
    print("Show Status:", status)


# Sample Input
total_seats = 200
booked_list = [1] * 150  # 150 booked seats

# Function call
seat_occupancy(total_seats, booked_list)
```

```python
#Problem 4: Cloud Server Load Classification System

def server_load(cpu_list):
    # Variable to store total CPU usage
    total = 0

    # Loop to calculate sum of CPU readings
    for i in range(len(cpu_list)):
        total = total + cpu_list[i]

    # Calculate average CPU load
    avg = total / len(cpu_list)

    # Classify server status based on average CPU load
    if avg < 50:
        status = "Normal"
    elif avg >= 50 and avg <= 80:
        status = "Warning"
    else:
        status = "Critical"

    # Print results
    print("Average CPU Load:", int(avg), "%")
    print("Server Status:", status)


# Sample Input
cpu_list = [45, 60, 70, 85, 90]

# Function call
server_load(cpu_list)
```

```python
#Problem 5: Smart Classroom Resource Usage Monitor

def classroom_usage(resources):
    # List to store names of overused resources
    overused = []

    # Define a simple usage limit (in hours)
    limit = 6

    # Loop through each resource and check usage
    for key in resources:
        if resources[key] > limit:
            overused.append(key)  # Add resource if usage exceeds limit

    # Check if any resource is overused
    if len(overused) > 0:
        alert = "Yes"
    else:
        alert = "No"

    # Print results
    print("Overused Resources:", overused)
    print("Energy Alert:", alert)


# Sample Input
resources = {
    "Projector": 6,
    "AC": 9,
    "Lights": 4
}

# Function call
classroom_usage(resources)
```

```python
#Problem 6: Online Event Registration Capacity Controller

def event_registration(capacity, registered):
    # Check if registrations exceed capacity
    if registered <= capacity:
        confirmed = registered       # All registrations can be confirmed
        waitlisted = 0
        status = "Open"
    else:
        confirmed = capacity         # Only capacity number can be confirmed
        waitlisted = registered - capacity  # Remaining go to waitlist
        status = "Closed"

    # Print final results
    print("Confirmed Registrations:", confirmed)
    print("Waitlisted Users:", waitlisted)
    print("Registration Status:", status)


# Sample Input
capacity = 100
registered = 105

# Function call
event_registration(capacity, registered)
```
